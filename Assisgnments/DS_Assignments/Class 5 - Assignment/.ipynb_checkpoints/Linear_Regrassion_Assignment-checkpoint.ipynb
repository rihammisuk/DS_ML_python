```python
import numpy as np
import pandas as pd

# Step 1: Create initial Excel file with dataset
weights = np.array([2, 4, 5, 3, 6, 5, 7])
prices = np.array([35, 60, 20, 50, 50, 55, 60])

# Create DataFrame for initial dataset
df_initial = pd.DataFrame({
    'Weight': weights,
    'Price': prices
})

# Save initial dataset to CSV
df_initial.to_csv("price_analysis.csv", index=False)
print("Initial CSV file 'price_analysis.csv' created successfully.")

# Step 2: Read Excel file and perform calculations

# Read the CSV file
df = pd.read_csv("price_analysis.csv")
weights = df['Weight'].to_numpy()
prices = df['Price'].to_numpy()

# Task 1: Your objective is to manually compute the slope (M) and y-intercept (C) using
#   Ordinary Least Squares Linear Regression. Once determined, apply these values to
#   predict the price when the vegetable weight is 6.

# Formula 1: Mean of X and Y
# Mean(X) = sum(x_i)/n, Mean(Y) = sum(y_i)/n
weights_mean = np.mean(weights)
prices_mean = np.mean(prices)

# Formula 2: Covariance of X and Y
# Cov(X, Y) = sum((x_i - x_mean)(y_i - y_mean))/n
numerator = np.sum((weights - weights_mean) * (prices - prices_mean))

# Formula 3: Variance of X
# Var(X) = sum((x_i - x_mean)^2)
denominator = np.sum((weights - weights_mean)**2)

# Formula 4: Slope (M)
# M = Cov(X, Y) / Var(X)
M = numerator / denominator
print(f"Slope (M): {M:.2f}")

# Formula 5: Intercept (C)
# C = y_mean - M * x_mean
C = prices_mean - M * weights_mean
print(f"Intercept (C): {C:.2f}")

# Formula 6: Predicted Price for a Given Weight
# y_hat = M * x + C
predicted_price = M * 6.50 + C
print(f"Predicted Price for weight 6.50: {predicted_price:.2f}")

# Task 2: Compute Residuals and Metrics

# Formula 7: Predicted Prices for All Data Points
# y_hat_i = M * x_i + C
predicted_prices = M * weights + C

# Formula 8: Residuals
# Residual_i = y_i - y_hat_i
residuals = prices - predicted_prices

# Print residuals for each data point
for i in range(len(weights)):
    print(f"Data Point {i+1}: Residual (Error) = {residuals[i]:.2f}")

# Formula 9: Mean Squared Error (MSE)
# MSE = sum(Residual_i^2)/n
mse = np.mean(residuals**2)
print(f"MSE: {mse:.2f}")

# Formula 10: Mean Absolute Error (MAE)
# MAE = sum(|Residual_i|)/n
mae = np.mean(np.abs(residuals))
print(f"MAE: {mae:.2f}")

# Final Task: Generate Updated Excel File (Including weight = 6.5)

# Formula 11: DataFrame Creation (Extended for weight = 6.5)
# DataFrame = Table with columns: {x_i, y_i, y_hat_i, Residual_i} plus row for {6.5, NaN, y_hat_6.5, NaN}
weights_extended = np.append(weights, 6.5)
prices_extended = np.append(prices, np.nan)  # No actual price for weight = 6.5
predicted_prices_extended = M * weights_extended + C
residuals_extended = prices_extended - predicted_prices_extended  # NaN for weight = 6.5

# Create DataFrame
df_prices = pd.DataFrame({
    'Weight': weights_extended,
    'Price': prices_extended,
    'Predicted Price': predicted_prices_extended,
    'Residual': residuals_extended
})

# Formula 12: Export to CSV
# Save DataFrame as CSV file (.csv)
df_prices.to_csv("price_analysis.csv", index=False)
print("Updated CSV file 'price_analysis.csv' created successfully.")
```
