1. Variable Swap: Write a Python program to swap the values of two variables without using a temporary variable.


```python
swapA = 10
swapB = 20
print("print before swapping : ")
print(swapA)
print(swapB)
swapA, swapB = swapB, swapA
print("print after swapping : ")
print(swapA)
print(swapB)
```

2. Even or Odd: Write a Python program that takes an integer as input and prints whether it is even or odd.

```python
evenoddInput = int(input("Enter an interger : "))
if(evenoddInput % 2 == 0):
    print(f"{evenoddInput} is Even.")
else:
    print(f"{evenoddInput} is odd.")
```

3. String Reverse: Write a Python function to reverse a given string and return the reversed string.

```python
def reverseStringMethod(simpleString):
    return simpleString[::-1]
simpleString = (input("Enter an string : "))
reverseString = reverseStringMethod(simpleString)
print(reverseString)
```

5. Temperature Converter: Write a Python program that converts a temperature in Celsius to Fahrenheit. Take the Celsius temperature as input from the user.

```python
celcius = float(input("Enter temparature in Celcius : "))
farenheit = (celcius * 9/5) + 32
print(f"{celcius} degree celcius = {farenheit} degree farenheit ")
```

6. Data Type Checker: Write a Python function that takes a variable as input and returns the data type of the variable as a string (e.g., “int”, “float”, “str”, “list”, etc.).

```python
def getPythonDataType(value):
    return type(value).__name__

print(getPythonDataType(5))
print(getPythonDataType(5.5))
print(getPythonDataType("five"))
print(getPythonDataType([1,2,3,4,5]))
```

7. String Palindrome: Write a Python function to check if a given string is a palindrome or not.

```python
def is_Palindrome(value):
    value = value.lower().replace(" ", "")
    return value == value[::-1]

print(is_Palindrome("riham"))
print(is_Palindrome("racecar"))
```

8. String Reversal with Slicing: Write a Python function to reverse a given string using slicing.

```python
def reverse_String_Slice(s):
    return s[::-1]
print(reverse_String_Slice("riham"))
```

9. String Concatenation: Write a Python program that takes two strings as input and concatenates them into a single string without using the + operator.

```python
concatString1 = input("Enter first String: ")
concatString2 = input("Enter second String: ")
concatStringResult = "".join([concatString1, concatString2])
print(concatStringResult)
```
